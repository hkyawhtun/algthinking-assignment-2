```python
from max_heap import MaxHeap

h = MaxHeap()

values = [10, 50, 25, 80, 30, 100, 5]

for value in values:
    h.push(value)

print("Heap:", h.array)
print("Top:", h.getTop())

results = []

while h.N > 0:
    results.append(h.pop())

print("Pop order:", results)
```
